```python
#Ridge and Lasso regreassion
#step 1
x8 = np.random.rand(100, 5)
y8 =  x8 @ np.array([5,4,3,2,1]) + np.random.randn(100)

#step 2: ridge regreassion
from sklearn.linear_model import Ridge

ridge = Ridge ( alpha=3.0)
ridge.fit(x8, y8 )
```

```python
#step 3 lasso
x8 = np.random.rand(100, 5)
y8 =  x8 @ np.array([5,4,3,2,1]) + np.random.randn(100)

#step 2: ridge regreassion
from sklearn.linear_model import Lasso

lasso = Lasso ( alpha=0.1)
lasso.fit(x8, y8 )
```

```python

#Visualiza coefficients
plt.bar(range(5),ridge.coef_,label="Ridge")
plt.bar(range(5),lasso.coef_,label="Lasso")
plt.legend()
plt.title("Ridge vs Lasso coefficients")
plt.show()
```

```python
from sklearn.svm import SVR
import numpy as np
import matplotlib.pyplot as plt
```

```python

#vector regression(SVR)
#step 1
x9 = np.linspace(-5,5,100).reshape(-1,1)
y9 = np.sin(x9).ravel()
#step 2 train svr


svr =SVR(kernel='rbf',C=100,gamma=0.1)
svr.fit(x9,y9)
y9_pred = svr.predict(x9)

```

```python
#step 3 visualize

plt.scatter(x9, y9, label="Actual Data")
plt.plot(x9, y9_pred, label="SVR Prediction")
plt.title("Support Vector Regression (RBF)")
plt.xlabel("X values")
plt.ylabel("sin(X)")
plt.legend()
plt.show()
```

```python
#Non-linear interpretable
#step 1p
X=np.sort(np.random.rand(80),axis=0)
Y=np.sin(5*X).ravel()
```

```python
#Step 2: Train Model
from sklearn.tree import DecisionTreeRegressor

tree =DecisionTreeRegressor(max_depth=3)

#Reshape x5 to be a 2D array
X_reshaped = X.reshape(-1,1)
tree.fit(X_reshaped,Y)
Y_pred = tree.predict(X_reshaped)
```

```python
#Step 3: Visualize
plt.scatter(X, Y, label='Actual Data')

plt.plot(X_reshaped, Y_pred, color='red', label='Decision Tree Prediction')

plt.xlabel('X')

plt.ylabel('Y')

plt.title('Decision Tree Regression')

plt.legend()

plt.show()
```
